## 1) Read csv file 2) load csv to bronze table

```python
%sql
drop table if exists productcatalog.bronze.product_bronze
```

```python
filepath = "/Volumes/productcatalog/productschema/filestorage/csv/products-1000.csv"
df = spark.read.format("csv")\
        .option("header", "true")\
        .option("inferSchema", "true")\
        .load(filepath)
df.withColumnRenamed("Internal ID", "InternalID")\
        .write.mode("overwrite")\
        .saveAsTable("productcatalog.bronze.product_bronze")
```

```python
%sql
select * from productcatalog.bronze.product_bronze
```

```python
filepath = "/Volumes/productcatalog/productschema/filestorage/csv/product-catalog-ai-100.csv"
df = spark.read.format("csv")\
        .option("header", "true")\
        .option("inferSchema", "true")\
        .load(filepath)

df.withColumnRenamed("Product ID", "product_id")\
        .withColumnRenamed("Product Title", "product_title")\
        .withColumnRenamed("Raw Description", "raw_description")\
        .withColumnRenamed("Target Language", "target_language")\
        .write.mode("overwrite")\
        .saveAsTable("productcatalog.bronze.product_ai_bronze")
```
